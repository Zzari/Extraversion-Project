## Background of the Study

Extraversion is a dimension of the Five Factor Model (FFM) and Myers-Briggs Type Indicator (MBTI) that serves as the leading personality model being used for assessing enduring patterns of human personality traits (McCrae, 2009). Although scientifically known to have a biological basis, like genetics, gregariousness is a major indicator that defines this spectrum. It can dictate how individuals interact with their social environments, including their talkativeness and excitability. Extroverts tend to be sociable, deriving energy from hanging out with others. They seek more exciting external experiences and have good teamwork skills (Fieri et al., 2023). Meanwhile, introverted people are more inward; they recharge through solitude and prefer working independently. Being self-reliant, they tend to be good listeners and are more empathetic and insightful.

It is important to study extraversion as it can predict effective functioning and well-being across cognitive performance and even economic status (Ozer & Benet-Martinez, 2006; Roberts, Kuncel, Shiner, Caspi, & Goldberg, 2007; Revelle, 2008). It also helps in predicting the risk for mental illnesses and how they respond to stress (Widiger, 2005). Thus, it can help us understand what we can offer and, at the same time, the support that we need, leading to a more strategic approach to our work. Furthermore, predicting this can be very valuable in various fields. In school, it is being used for career counseling. In marketing, it can shed light on the users’ generated content (UGC) to understand the tendencies of the consumers. Previous studies have used machine learning models to train datasets to predict personality traits. Some methods used were K-Nearest Neighbour, Decision Tree, and Logistic Regression. (Fieri et al., 2023).

### **📈 Model Evaluation Metrics**
In this study, we applied supervised classification models (K-Nearest Neighbors and Logistic Regression) to predict behavioral and psychological traits based on social interaction patterns. To evaluate the performance and reliability of our models, we used several key evaluation metrics: accuracy, precision, recall, and the F1-score. These metrics are especially relevant given the nature of our hypotheses, which involve behavioral traits that may not be equally distributed in the population.


| Metric     | Explanation                                                   | Rationale                                                                 |
|------------|---------------------------------------------------------------|---------------------------------------------------------------------------|
| **Accuracy**  | Proportion of total correct predictions out of all predictions. | Gives a general sense of model performance but can be misleading on imbalance. |
| **Precision** | Proportion of true positives out of all predicted positives.     | Useful when false positives are costly (e.g., mislabeling an introvert as extrovert). |
| **Recall**    | Proportion of true positives out of all actual positives.       | Important when false negatives are costly (e.g., missing an extroverted profile). |
| **F1-Score**  | Harmonic mean of precision and recall.                          | Balances precision and recall — especially useful with imbalanced datasets. |

**Relevance to Study**
* **Accuracy**: Accuracy gives a broad overview of how well the model classifies both classes — for example, predicting whether someone is drained after socializing (Drained_Flag = 1) or not. However, in a dataset where one class dominates (e.g., most people are not drained), accuracy can be misleading. Hence, we also use other metrics.

* **Precision**: If our model predicts that someone is drained by socializing, precision tells us how often that prediction is actually correct. This is important when providing personalized recommendations — we want to avoid wrongly labeling someone as socially drained when they’re not.

* **Recall**: High recall ensures that we identify most of the individuals who actually feel drained or exhibit the trait in question. This is especially important in behavioral studies where false negatives (missing real cases) could lead to overlooking people who might need support or different engagement strategies.

* **F-1 Score**: Because our study involves subtle, overlapping behavioral traits (not clear-cut labels), F1-score is an excellent metric. It helps us strike a balance between not overpredicting a trait (precision) and not missing too many real cases (recall). This is especially useful in classifying nuanced psychological conditions like social fatigue.

In this study, we applied two supervised classification algorithms — **K-Nearest Neighbors (KNN)** and **Logistic Regression (LR)** — to test our hypotheses about social behavior patterns such as stage fear and social exhaustion. ### 🔸 K-Nearest Neighbors (KNN)

KNN is a distance-based, non-parametric algorithm that classifies a data point based on the majority label among its **k nearest neighbors** in the feature space.

#### ✔ How it works:
1. For a given input, calculate the distance (usually Euclidean) to all training data points.
2. Identify the `k` nearest neighbors (we experimented with values like `k = 3, 5, 7`).
3. Assign the class that appears most frequently among those neighbors.

#### 📌 Why we used KNN:
- It’s intuitive and works well when the data is already normalized (which we ensured during preprocessing).
- It’s useful for capturing **non-linear decision boundaries** which may arise from social-behavior traits.
- We could easily interpret **how local feature similarity affects class predictions**, which is helpful in understanding behavioral clusters.

#### ⚙ Key steps in our notebook:
1. Scaled relevant features using Min-Max Scaling.
2. Split data into training and testing sets.
3. Used `KNeighborsClassifier` from scikit-learn.
4. Tested multiple values for `k` and selected the one with the best F1-score.
5. Evaluated performance using classification metrics like accuracy, precision, recall, and F1-score.


### 🔸 Logistic Regression (LR)

Logistic Regression is a linear model used for binary classification that estimates the probability that a given input belongs to a certain class using the logistic (sigmoid) function.

#### ✔ How it works:
1. The model calculates a **weighted sum of input features**, adds a bias term, and passes the result through a sigmoid function:
  $$ P(y = 1) = \frac{1}{1 + e^{-(\beta_0 + \beta_1x_1 + \dots + \beta_nx_n)}} $$
   
2. If the predicted probability is greater than a threshold (default is 0.5), it assigns the positive class.

#### 📌 Why we used LR:
- It provides interpretable coefficients, helping us understand **how each feature (e.g., time spent alone or friends circle size) influences social behavior outcomes**.
- It's computationally efficient and performs well when the relationship between features and target is roughly linear.
- It is a common baseline model for binary classification.

#### ⚙ Key steps in our notebook:
1. Used `LogisticRegression` from scikit-learn.
2. Trained the model using the same train/test splits as KNN.
3. Evaluated model using the same metrics (accuracy, precision, recall, F1-score).
4. Interpreted the model’s coefficients to assess which features were most predictive.

# Feature Selection + Hyperparameter Tuning

### 🔧 Hyperparameter Tuning Using GridSearchCV

To fine-tune our models and improve predictive performance, we used **GridSearchCV** from `sklearn.model_selection`. GridSearch performs **exhaustive search over specified hyperparameter combinations**, using **cross-validation (CV)** to evaluate each setting.

---

#### 🟦 K-Nearest Neighbors (KNN)

We used a grid search to optimize the `n_neighbors` parameter, which controls how many neighbors the algorithm considers when assigning class labels.

---

#### 🟦 Logistic Regression (LogReg)

We used a grid search to optimize the `C` parameter, which controls the strength of the regularization of the features, `penalty` parameter, which chooses whether to use ridge or lasso regularization, and `solver` parameter, which utilizes `liblinear` solver to find the best weights that the algorithm considers to predict values.


```python
#For hyperparameter Tuning:
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
```

```python
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split


def tune_and_evaluate(X, y, model_type='knn', model_name='Model', param_grid=None):

    # Default param grids if none provided
    if model_type == 'knn':
        model = KNeighborsClassifier()
        default_grid = {
            'model__n_neighbors': list(range(1, 11)),
            'model__weights': ['uniform', 'distance'],
            'model__p': [1, 2] #p = 1 (manhattan distance), p = 2(euclidean distance)
        }
    elif model_type == 'logreg':
        model = LogisticRegression(max_iter=1000)
        default_grid = {
            'model__C': [0.01, 0.1, 1, 10],
            'model__penalty': ['l1','l2'],
            'model__solver': ['liblinear']
        }
    else:
        raise ValueError("model_type must be 'knn' or 'logreg'")

    # Override default grid if custom grid provided
    grid_to_use = param_grid if param_grid else default_grid

    # Create pipeline with scaler and model
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('model', model)
    ])

    # Train-test split
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

    # Grid search
    grid = GridSearchCV(pipe, grid_to_use, cv=5)
    grid.fit(X_train, y_train)

    # Evaluation
    print(f"🔍 {model_name.upper()} ({model_type.upper()})")
    print("Best Parameters:", grid.best_params_)
    print("Classification Report:")
    print(classification_report(y_test, grid.predict(X_test)))
    print("Train accuracy:", grid.best_estimator_.score(X_train, y_train))
    print("Test accuracy:", grid.best_estimator_.score(X_test, y_test))


    return grid.best_estimator_
```

```python
hypotheses = {
    "H1": {
        "X": final_df[['Post_frequency', 'Time_spent_Alone', 'Going_outside', 'Friends_circle_size', 'Drained_flag']],
        "y": final_df['High_Attendance']
    },
    "H2": {
        "X": final_df[['Time_spent_Alone', 'Social_event_attendance', 'Friends_circle_size', 'Drained_flag']],
        "y": final_df['Stage_fear']
    },
    "H3": {
        "X": final_df[['Social_event_attendance', 'Going_outside', 'Post_frequency', 'Stage_fear', 'Drained_flag']],
        "y": final_df['Friend_Class']
    },
    "H4": {
        "X": final_df[['Friends_circle_size', 'Social_event_attendance', 'Time_spent_Alone', 'Drained_flag']],
        "y": final_df['High_Poster']
    }
}
```

```python
for hypo_key, data in hypotheses.items():
    for model_type in ['knn', 'logreg']:
        tune_and_evaluate(
            X=data['X'],
            y=data['y'],
            model_type=model_type,
            model_name=f"{hypo_key} - {model_type}"
        )
```
